```python
df = pd.read_csv("heart disease.csv")
```

```python
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df.select_dtypes(exclude=[np.number]).columns.tolist()
```

```python
for col in num_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

for col in cat_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing values handled. Current missing count:", df.isnull().sum().sum())
```

```python
df_encoded = pd.get_dummies(df, columns=cat_cols, drop_first=True)

print("Encoded Dataset Shape:", df_encoded.shape)
display(df_encoded.head())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
features_to_scale = [col for col in num_cols if col in df_encoded.columns and len(df_encoded[col].unique()) > 5]

df_encoded[features_to_scale] = scaler.fit_transform(df_encoded[features_to_scale])

print("Scaled features:", features_to_scale)
display(df_encoded.head())
```

```python
from sklearn.ensemble import IsolationForest
iso = IsolationForest(contamination=0.02, random_state=42)
outliers = iso.fit_predict(df_encoded)


df_encoded['is_outlier'] = outliers

num_outliers = (df_encoded['is_outlier'] == -1).sum()
print(f"Identified {num_outliers} outliers (5% contamination rate).")


display(df_encoded[df_encoded['is_outlier'] == -1].head())
```
